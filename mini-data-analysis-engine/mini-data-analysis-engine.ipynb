### PROJECT 10: Mini Data Analysis Engine


Build a system that:

Takes a dataset  
Computes key statistics  
Detects outliers


```python
import numpy as np

# User input
user_input = input("Enter numbers separated by commas: ")

# Convert to array
data = np.array([float(x) for x in user_input.split(",")])

# Analysis function
def analyze_data(data):
    mean = np.mean(data)
    median = np.median(data)
    std = np.std(data)

    z_scores = (data - mean) / std
    outliers = data[np.abs(z_scores) > 2]

    print("\nResults:")
    print("Mean:", mean)
    print("Median:", median)
    print("Standard Deviation:", std)
    print("Outliers:", outliers)

# Run analysis
analyze_data(data)


### Breakdown:

# .split(",") → turns string into list
# float(x) → converts each value to number
# np.array() → turns it into NumPy array
```
