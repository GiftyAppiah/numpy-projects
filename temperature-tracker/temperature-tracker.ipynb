### Project 2(Temperaure)

```python
import numpy as np

temp_celsius = np.array([30,32,31,29,35,36,33])

## converting celsius to fahrenheit
temp_fahren = (temp_celsius * 1.8) +32
print(temp_fahren.astype(int))


## Hottest day
hottest = temp_fahren.max()
print('the hottest temp is',hottest)

## coldest day
coldest = temp_fahren.min()
print('the coldest temp is',coldest)
```
