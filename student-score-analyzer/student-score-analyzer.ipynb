### Project 1(Student Score Analyzer)

```python
import numpy as np
import pandas as pd


scores = np.array([45,67,89,56,72])  
print(scores)


## Finding mean
average = scores.mean()
print(average)

## Highest number
highest_num = scores.max()
print(highest_num)

## Lowest score
lowest_num = scores.min()
print(lowest_num)


## Finding number scores greater than average
x = (scores > average)
print(x)
count = np.sum(x)
print(count)
```
