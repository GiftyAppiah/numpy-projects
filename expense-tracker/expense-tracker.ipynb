### Project 3: Simple Expense Tracker

```python
import numpy as np
expenses = np.array([120, 50, 300, 80, 200])
print(expenses)

## Total spending
total = expenses.sum()
print('Total spending is',total)

## Average spending
average = expenses.mean()
print('Average spending is',average)

## Expenses greater than 100
x = expenses > 100
count = x.sum()
print(count)
print(x)

if (expenses > 100).any:
    print(expenses)
```
