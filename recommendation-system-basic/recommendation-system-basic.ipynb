### Project 11: Recommendation Logic (Starter)    

**Concepts:**  

Similarity calculations  
**Idea:**     
Compare user preferences using arrays    
Recommend based on similarity 

```python
import numpy as np

ratings = np.array([
    [5, 3, 0, 1],
    [4, 0, 0, 1],
    [1, 1, 0, 5],
    [0, 0, 5, 4]
])

user_index = 0
user = ratings[user_index]

# Similarity
similarity = ratings @ user

# Ignore self
similarity[user_index] = -1

# Find most similar user
most_similar_user = np.argmax(similarity)

# Get their ratings
recommended = ratings[most_similar_user]

# Recommend unseen items with high ratings
recommendations = np.where((user == 0) & (recommended > 3))

print("Recommended item indices:", recommendations[0])
```
