```python
import numpy as np

scores = np.array([45, 67, 89, 56, 72, 91, 38])

Grade = np.where(scores >= 80, 'A',
        np.where((scores >= 60) & (scores < 80), 'B' ,   
        np.where((scores >= 50) & (scores < 60), 'C',
        np.where(scores < 50, 'D', 'F'))))

print(Grade)

print ('A :', np.sum( Grade == 'A'))
print ('B :', np.sum( Grade == 'B'))
print ('C :', np.sum( Grade == 'C'))
print ('D :', np.sum( Grade == 'D'))




### Lesson: where is a condiion statement that has 3 arguments
##It ALWAYS needs 3 things:

# 1. Condition
# 2. What to return if True
# 3. What to return if False

# Condiions are not supposed to have list tag [] inside the
```
