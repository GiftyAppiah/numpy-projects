```python
import numpy as np

### Generating a 3 by 3 matrix randomly from range 1-10
C =np.arange(1,10).reshape(3,3)
print(C)

### Generating a 3 by 3 matrix randomly from range 1-10
D = np.arange(1,10).reshape(3,3)
print(D)

### Smming the two matrix
matrix_sum = D + C
print('Sum:', matrix_sum)

### Multiplying the two matrix

matrix_mul = D*C
print('Multiplication : ' ,matrix_mul)

## transposeing D
D_Transpose = np.transpose(D)
print('Transpose:', D_Transpose)
```
