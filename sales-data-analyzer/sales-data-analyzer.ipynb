### Project 7: Sales Data Analyzer


```python
#### sales data for 3 products across 5 days

import numpy as np
sales = np.array([
    [200, 220, 210, 205, 230],
    [150, 160, 170, 165, 180],
    [300, 310, 305, 290, 320]
])


### total sales per product
product_1 = np.sum(sales[0,:])
print('Total sales for P1:', product_1)

product_2 = np.sum(sales[1,:])
print('Total sales for P2:', product_2)

product_3 = np.sum(sales[2,:])
print('Total sales for P3:', product_3)

### Most efficient way
total_per_product = np.sum(sales, axis=1)
print(total_per_product)



#### total per day
day_1 = np.sum(sales[:,0])
print('Total sales for D1:', day_1)

day_2 = np.sum(sales[:,1])
print('Total sales for D2:', day_2)

day_3 = np.sum(sales[:,2])
print('Total sales for D3:', day_3)


day_4 = np.sum(sales[:,3])
print('Total sales for D4:', day_4)


day_5 = np.sum(sales[:,4])
print('Total sales for D5:', day_5)


### Most efficient way
total_per_day = np.sum(sales, axis=0)
print(total_per_day)




## Best selling product

best_product = total_per_product.max()
print('Best selling product:', best_product)
```
