### Project 5: Data Normalization Tool  
Task:

Normalize data using:

```python
import numpy as np

scores = np.array([45, 67, 89, 56, 72, 91, 38])

average = scores.mean()
print(average)

standard_deviaion = scores.std()
print (standard_deviaion)

normalization = (scores - average) / standard_deviaion
print(normalization)
```
